##Setup the project environment for the Formula-1 Incremental Load Project
1. Create an external location databricks-course-ext-dl1-formula1-incr
2. Create Catalog i.e. formula1_incr
3. Create schemas for Landing, Bronze, Silver, Gold
4. Create Volume files in the Landing Schema

###Access the Cloud Storage

```sql
%fs ls 'abfss://formula1-incr-load@databrickscourse1dl.dfs.core.windows.net/landing/'
```

###1. Create External Location

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS databricks_course_ext_dl1_formula1_incr_load
URL 'abfss://formula1-incr-load@databrickscourse1dl.dfs.core.windows.net'
WITH (STORAGE CREDENTIAL `databricks-course-sc`)
COMMENT 'External Location for Formula-1 Incremental Load Project Container';
```

### 2. Create Catalog i.e. Formula1

```sql
SHow CATALOGS;
```

```sql
CREATE CATALOG IF NOT EXISTS formula1_incr
   MANAGED LOCATION 'abfss://formula1-incr-load@databrickscourse1dl.dfs.core.windows.net'
   COMMENT 'This is the catalog of the formula1 incremental load project';
```

### 3. Create Schemas for Landing, Bronze, Silver, Gold

```sql
CREATE SCHEMA IF NOT EXISTS formula1_incr.landing;
CREATE SCHEMA IF NOT EXISTS formula1_incr.bronze
    MANAGED LOCATION 'abfss://formula1-incr-load@databrickscourse1dl.dfs.core.windows.net/bronze';
CREATE SCHEMA IF NOT EXISTS formula1_incr.silver
    MANAGED LOCATION 'abfss://formula1-incr-load@databrickscourse1dl.dfs.core.windows.net/silver';
CREATE SCHEMA IF NOT EXISTS formula1_incr.gold
    MANAGED LOCATION 'abfss://formula1-incr-load@databrickscourse1dl.dfs.core.windows.net/gold';
```

### 4. Create Volume named Files

```sql
CREATE EXTERNAL VOLUME formula1_incr.landing.files
LOCATION 'abfss://formula1-incr-load@databrickscourse1dl.dfs.core.windows.net/landing';
```
