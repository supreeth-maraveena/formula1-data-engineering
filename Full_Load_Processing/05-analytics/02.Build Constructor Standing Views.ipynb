##Build Constructor Standings
####Sources
1. fact_session_results
2. dim_constructors

####Output Columns
1. season
2. constructor id
3. constructor name
4. nationality
5. race starts
6. total points
7. number of wins
8. number of podiums
9. standing position

```sql
CREATE OR REPLACE VIEW formula1.gold.v_constructor_standings
AS 
WITH constructor_session_summary
AS
(SELECT r.season,
        c.constructor_id,
        c.constructor_name,
        c.nationality,
        COUNT(*) AS race_starts,
        SUM(r.points) AS total_points,
        count_if(r.is_win) AS number_of_wins,
        count_if(r.is_podium) AS number_of_podiums

    FROM formula1.gold.fact_session_results r
    JOIN formula1.gold.dim_constructors c
         ON r.constructor_id = c.constructor_id

    GROUP BY r.season,
             c.constructor_id,
             c.constructor_name,
             c.nationality)

SELECT season,
       constructor_id,
       constructor_name,
       nationality,
       RANK() OVER (PARTITION BY season ORDER BY total_points DESC, number_of_wins DESC) AS standings,
       race_starts,
       total_points,
       number_of_wins,
       number_of_podiums
FROM constructor_session_summary;
```

```sql
SELECT * FROM formula1.gold.v_constructor_standings WHERE season = 2025;
```
