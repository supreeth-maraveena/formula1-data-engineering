##Setup the project environment for the Formula-1 Project
1. Create an external location databricks-course-ext-dl1-formula1
2. Create Catalog i.e. Formula1
3. Create schemas for Landing, Bronze, Silver, Gold
4. Create Volume files in the Landing Schema

###Access the Cloud Storage

```sql
%fs ls 'abfss://formula1@databrickscourse1dl.dfs.core.windows.net/landing/'
```

###1. Create External Location

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS databricks_course_ext_dl1_formula1
URL 'abfss://formula1@databrickscourse1dl.dfs.core.windows.net'
WITH (STORAGE CREDENTIAL `databricks-course-sc`)
COMMENT 'External Location for Formula-1 Project Container';
```

### 2. Create Catalog i.e. Formula1

```sql
SHow CATALOGS;
```

```sql
CREATE CATALOG IF NOT EXISTS Formula1
   MANAGED LOCATION 'abfss://formula1@databrickscourse1dl.dfs.core.windows.net'
   COMMENT 'This is the main catalog of the formula1 project';
```

### 3. Create Schemas for Landing, Bronze, Silver, Gold

```sql
CREATE SCHEMA IF NOT EXISTS formula1.landing;
CREATE SCHEMA IF NOT EXISTS formula1.bronze
    MANAGED LOCATION 'abfss://formula1@databrickscourse1dl.dfs.core.windows.net/bronze';
CREATE SCHEMA IF NOT EXISTS formula1.silver
    MANAGED LOCATION 'abfss://formula1@databrickscourse1dl.dfs.core.windows.net/silver';
CREATE SCHEMA IF NOT EXISTS formula1.gold
    MANAGED LOCATION 'abfss://formula1@databrickscourse1dl.dfs.core.windows.net/gold';
```

From the below commands, we can see that we are not in the correct catalog i.e. formula1 and hence only seeing 2 schemas.

We have to change the catalog to see the schemas we require.

```sql
SELECT current_catalog();
```

```sql
SHOW SCHEMAS;
```

Now we shall see the correct schemas

```sql
USE CATALOG formula1;
```

```sql
SELECT current_catalog();
```

```sql
SHOW SCHEMAS;
```

### 4. Create Volume named Files

```sql
CREATE EXTERNAL VOLUME IF NOT EXISTS formula1.landing.files
LOCATION 'abfss://formula1@databrickscourse1dl.dfs.core.windows.net/landing';
```
